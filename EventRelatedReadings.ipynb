## Basic lexicon

```python
%%lamb reset, ambiguity
# plurality
# pl = L g_<e,t>: L f_<e,t>: Forall z_e: f(z) ==> g(z)
# plplus = L g_<e,t>: L f_<e,t>: pl(g)(f) & Card_<<e,t>,n>(f)>0
pl = Pow_<<e,t>,<<e,t>,t>>
plplus = PowX_<<e,t>,<<e,t>,t>>
||PL|| = pl
||s|| = plplus
# union = L p_<<e,t>,t>: L y_e: Exists g_<e,t>: p(g) & g(y)
union = U_<<<e,t>,t>,<e,t>>
# existential closure
||EC|| = L f_<v,t>: Exists e_v: f(e)
# theta roles
||THEME|| = L y_e: L f_<v,t>: L e_v: f(e) & (Exists x_e: Substage_<(e,e),t>(x,y) & Theme_<(e,v),t>(x,e))
# contextual domain restriction
||C|| = L x_e: True # trivial
||C|| = L x_e: Exists e_v: PassThruLock_<v,t>(e) & Theme_<(e,v),t>(x,e) # being the theme of a passing event
# maximality
# maX = L u_(e,<e,t>): u[1](u[0]) & (Forall y_e: (u[1](y) & Substage_<(e,e),t>(u[0],y)) ==> (u[0]==y))
maX = Max_<(e,<e,t>),t>
# everything else!
||passed|| = L e_v: PassThruLock_<v,t>(e)
||last_year|| = L f_<v,t>: L e_v: f(e) & LastYear_<v,t>(e)
||ship|| = Ship_<e,t>
||ships|| = Ships_<<e,t>,t>
#||four_thousand|| = L c_<e,t>: L p_<<e,t>,t>: L q_<<e,t>,t>: Exists f_<e,t>: (Forall x_e: f(x) ==> maX(x,L y_e: (Exists g_<e,t>: p(g) & g(y)) & c(y))) & Card_<<e,t>,n>(f)==4000 & q(f)
||four_thousand|| = L c_<e,t>: L p_<<e,t>,t>: L q_<<e,t>,t>: Exists f_<e,t>: (Forall x_e: f(x) ==> maX(x,L y_e: union(p)(y) & c(y))) & Card_<<e,t>,n>(f)==4000 & q(f)
||different|| = L p_<<e,t>,t>: plplus(L z_e: maX(z,union(p)))
```

Notational equivalents between formulae in this notebook and the paper:

$$\begin{align*}
Pow&:=\wp\\
PowX&:=\wp^{+}\\
U&:=\bigcup\\
Card(X)&:=\left|X\right|\\
Substage(x,y)&:=x\sqsubseteq y\\
Max&:=\mbox{max}^{\sqsubseteq}\\
Z&:=\zeta
\end{align*}$$

The commented-out sections in the lexicon above contain logical definitions of $\wp$, $\wp^{+}$, $\mbox{max}^{\sqsubseteq}$ and $\bigcup$. It turns out that in the representation of formulae these make things _less_ clear, so I've left them undefined in the end.

```python
counter = 0
def eg(example:str, counter:int) -> None:
    print(f'({counter})\t{example}\n')

def exemplify(example:str, counter:int, compResult) -> None:
    eg(example, counter)
    if len(compResult)<1:
        raise Exception("Composition failed for some reason.")
    print(compResult.source)
    print("")
    if len(compResult)<2:
        display(compResult.content[0].content)
    else:
        for n,cont in enumerate(compResult.content):
            print(f"Result {n+1}:")
            display(cont.content)
```

## Ships and different ships

```python
counter += 1
exemplify("Four thousand ships passed through the lock last year.", counter, ((((four_thousand * C) * (ship * s)) * (PL * (b[1] * (EC * ((t[1] * THEME) * (passed * last_year))))))))
```

**Result 1** is the object-related reading &mdash; counting maximal ship stages, i.e. ships.

$$\begin{align*}
\textit{Max}\left(x,\left(\lambda y_e.\bigcup\left(\wp^{+}\left(\textit{Ship}\right)\right)(y)\right)\right)&=\textit{Max}\left(x,\left(\lambda y_e.\textit{Ship}(y)\right)\right)\\
&=\textit{Max}\left(x,\textit{Ship}\right)
\end{align*}$$

**Result 2** is the event-related reading &mdash; counting maximal [ship that is the theme of an event of passing through the lock] stages, i.e. lock traversals by ships.

$$\begin{align*}
\textit{Max}\left(x,\lambda y_e.\bigcup\left(\wp^{+}(\textit{Ship})\right)(y)\wedge\left(\exists e_v.\textit{PassThruLock}(e)\wedge\textit{Theme}(y,e)\right)\right)&=\textit{Max}\left(x,\lambda y_e.\textit{Ship}(y)\wedge\left(\exists e_v.\textit{PassThruLock}(e)\wedge\textit{Theme}(y,e)\right)\right)
\end{align*}$$

```python
counter += 1
exemplify("Four thousand different ships passed through the lock last year.", counter, ((((four_thousand * C) * (different * (ship * s))) * (PL * (b[1] * (EC * ((t[1] * THEME) * (passed * last_year))))))))
```

**Result 1** is the object-related reading again:

$$\begin{align*}
\textit{Max}\left(x,\left(\lambda y_e.\bigcup\left(\wp^{+}\left(\lambda z_e.\textit{Max}\left(z,\bigcup\left(\wp^{+}(\textit{Ship})\right)\right)\right)\right)(y)\right)\right)&=\textit{Max}\left(x,\left(\lambda y_e.\bigcup\left(\wp^{+}\left(\lambda z_e.\textit{Max}\left(z,\textit{Ship}\right)\right)\right)(y)\right)\right)\\
&=\textit{Max}\left(x,\left(\lambda y_e.\left(\lambda z_e.\textit{Max}\left(z,\textit{Ship}\right)\right)(y)\right)\right)\\
&=\textit{Max}\left(x,\left(\lambda y_e.\textit{Max}\left(y,\textit{Ship}\right)\right)\right)\\
&=\textit{Max}(x,\textit{Ship})
\end{align*}$$

**Result 2** is incoherent:

$$\begin{align*}
\textit{Max}\left(x,\left(\lambda y_e.\bigcup\left(\wp^{+}\left(\lambda z_e.\textit{Max}\left(z,\bigcup\left(\wp^{+}(\textit{Ship})\right)\right)\right)\right)(y)\wedge\left(\exists e_v.\textit{PassThruLock}(e)\wedge\textit{Theme}(y,e)\right)\right)\right)&=\textit{Max}\Big(x,\big(\lambda y_e.\textit{Max}(y,\textit{Ship})\wedge\left(\exists e_v.\textit{PassThruLock}(e)\wedge\textit{Theme}(y,e)\right)\big)\Big)\\
&=\textit{Max}(x,\emptyset)
\end{align*}$$

Therefore, only the object-related reading is available.

```python
%%lamb
# contextual domain restriction
||C[*]|| = L x_e: Exists y_e: Exists e_v: Substage_<(e,e),t>(y,x) & PassThruLock_<v,t>(e) & Theme_<(e,v),t>(y,e) # having a substage that is the theme of a passing event
||most|| = L c_<e,t>: L p_<<e,t>,t>: L q_<<e,t>,t>: 2 * Card_<<e,t>,n>(union(L f_<e,t>: pl(L x_e: maX(x, L y_e: union(p)(y) & c(y)))(f) & q(f))) > Card_<<e,t>,n>(L x_e: maX(x, L y_e: union(p)(y) & c(y)))
||at_night|| = L f_<v,t>: L e_v: f(e) & AtNight_<v,t>(e)
```

There's a typo in the paper in the lexical entry for _most_: it has the comparison sets the wrong way around. This is corrected in the form given above.

```python
counter +=1
exemplify("Most ships passed through the lock at night.", counter, ((((most * C) * (ship * s)) * (PL * (b[1] * (EC * ((t[1] * THEME) * (passed * at_night))))))))
```

**Result 1** is the basic object-related reading, counting maximal ship stages, i.e. ships:

$$\begin{align*}
Max(x,\lambda y.\bigcup(\wp^{+}(Ship))(y))&=Max(x,\lambda y.Ship(y))\\
&=Max(x,Ship)
\end{align*}$$

**Result 2** is the event-related reading, &lsquo;most lock traversals happened at night&rsquo;, counting maximal [ship that is the theme of a lock traversal] stages, i.e. lock traversals by ships:

$$
Max\left(x,\lambda y.\bigcup(\wp^{+}(Ship))(y)\wedge\exists e.PassThruLock(e)\wedge Theme(y,e)\right)=Max\left(x,\lambda y.Ship(y)\wedge\exists e.PassThruLock(e)\wedge Theme(y,e)\right)
$$

**Result 3** is the object-related reading with contextual domain restricion, &lsquo;most ships _that passed through the lock_ did so at night&rsquo;, counting maximal [ship stages that have a substage that is the theme of a lock traversal], i.e. ships that have traversed the lock:

$$
Max\left(x,\lambda y.\bigcup(\wp^{+}(Ship))(y)\wedge\exists e.\exists y'.PassThruLock(e)\wedge Substage(y',y)\wedge Theme(y'e)\right)=Max\left(x,\lambda y.Ship(y)\wedge\exists e.\exists y'.PassThruLock(e)\wedge Substage(y',y)\wedge Theme(y'e)\right)
$$

## Eventive nouns

```python
%%lamb
||passenger|| = L x_e: Exists e_v: Travel_<v,t>(e) & Theme_<(e,v),t>(x,e)
||Z|| = L f_<e,t>: L y_e: Exists x_e: (Substage_<(e,e),t>(x,y) | Substage_<(e,e),t>(y,x)) & f(x)
||fifty_million|| = L c_<e,t>: L p_<<e,t>,t>: L q_<<e,t>,t>: Exists f_<e,t>: (Forall x_e: f(x) ==> maX(x,L y_e: union(p)(y) & c(y))) & Card_<<e,t>,n>(f)==50000000 & q(f)
||served_meal|| = L e_v: MealService_<v,t>(e)
# new contextual domain restriction
||C|| = L x_e: True
||C[*]|| = L x_e: Exists e_v: Travel_<v,t>(e) & Theme_<(e,v),t>(x,e)
||C[*]|| = L x_e: Exists e_v: MealService_<v,t>(e) & Theme_<(e,v),t>(x,e)
```

### Passengers

#### Without $\zeta$:

```python
counter += 1
exemplify("Fifty million passengers were served a meal last year.", counter, ((((fifty_million * C) * (passenger * s)) * (PL * (b[1] * (EC * ((t[1] * THEME) * (served_meal * last_year))))))))
```

**Results 1&2** are both the event-related reading, counting persons-on-flights:

$$\begin{align*}
Max\left(x,\lambda y.\bigcup(\wp^{+}(\lambda x.\exists e.Theme(x,e)\wedge Travel(e)))(y)\right)&=Max\left(x,\lambda y.\bigcup(\wp^{+}(\lambda x.\exists e.Theme(x,e)\wedge Travel(e)))(y)\wedge\exists e.Theme(y,e)\wedge Travel(e)\right)\\
&=Max\left(x,\lambda y.\exists e.Theme(y,e)\wedge Travel(e)\right)
\end{align*}$$

**Result 3** is incoherent:

$$\begin{align*}
Max\left(x,\lambda y.\bigcup(\wp^{+}(\lambda x.\exists e.Theme(x,e)\wedge Travel(e)))(y)\wedge\exists e.MealService(e)\wedge Theme(y,e)\right)&=Max\left(x,\lambda y.\exists e.Theme(y,e)\wedge Travel(e)\wedge\exists e.MealService(e)\wedge Theme(y,e)\right)\\
&=Max(x,\emptyset)
\end{align*}$$

Therefore, without $\zeta$, only the counting persons-on-flights reading of (4) is available.

#### With $\zeta$:

```python
exemplify("Fifty million passengers were served a meal last year.", counter, ((((fifty_million * C) * ((Z * passenger) * s)) * (PL * (b[1] * (EC * ((t[1] * THEME) * (served_meal * last_year))))))))
```

**Result 1** is the object-related reading, counting people:

$$\begin{align*}
Max\left(x,\lambda y.\bigcup\left(\wp^{+}(\lambda y.\exists x.(\exists e.Theme(x,e)\wedge Travel(e))\wedge(x\sqsubseteq y\vee y\sqsubseteq x))\right)(y)\right)&=Max\left(x,\lambda y.\exists x.(\exists e.Theme(x,e)\wedge Travel(e))\wedge(x\sqsubseteq y\vee y\sqsubseteq x)\right)\\
&=Max(x,\lambda y.\exists z.z\sqsubseteq y \wedge\exists e.Travel(e)\wedge Theme(z,e))
\end{align*}$$

**Result 2** is the event-related reading that you get without $\zeta$:

$$\begin{align*}
Max\left(x,\lambda y.\bigcup\left(\wp^{+}(\lambda y.\exists x.(\exists e.Theme(x,e)\wedge Travel(e))\wedge(x\sqsubseteq y\vee y\sqsubseteq x))\right)(y)\wedge\exists e.Theme(y,e)\wedge Travel(e)\right)&=Max\left(x,\lambda y.\exists x.(\exists e.Theme(x,e)\wedge Travel(e))\wedge(x\sqsubseteq y\vee y\sqsubseteq x)\wedge\exists e.Theme(y,e)\wedge Travel(e)\right)\\
&=Max\left(x,\lambda y.\exists e.Theme(y,e)\wedge Travel(e)\right)
\end{align*}$$

**Result 3** is the fine-grained event-related reading, counting meals served to people on flights:

$$\begin{align*}
Max\left(x,\lambda y.\bigcup\left(\wp^{+}(\lambda y.\exists x.(\exists e.Theme(x,e)\wedge Travel(e))\wedge(x\sqsubseteq y\vee y\sqsubseteq x))\right)(y)\wedge\exists e.MealService(e)\wedge Theme(y,e)\right)&=Max\left(x,\lambda y.\exists x.(\exists e.Theme(x,e)\wedge Travel(e))\wedge(x\sqsubseteq y\vee y\sqsubseteq x)\wedge\exists e.MealService(e)\wedge Theme(y,e)\right)\\
&=Max\left(x,\lambda y.\exists x.(\exists e.Theme(x,e)\wedge Travel(e))\wedge y\sqsubseteq x\wedge\exists e.MealService(e)\wedge Theme(y,e)\right)
\end{align*}$$

### Different passengers

#### Without $\zeta$

```python
counter += 1
exemplify("Fifty million different passengers were served a meal last year.", counter, ((((fifty_million * C) * (different * (passenger * s))) * (PL * (b[1] * (EC * ((t[1] * THEME) * (served_meal * last_year))))))))
```

**Results 1&2** are both the event-related reading, counting persons-on-flights:

$$\begin{align*}
Max\left(x,\lambda y.\bigcup\left(\wp^{+}\left(\lambda z.Max\left(z,\bigcup\left(\wp^{+}(\lambda x.\exists e.Theme(x,e)\wedge Travel(e))\right)\right)\right)\right)(y)\right)&=Max\left(x,\lambda y.\bigcup\left(\wp^{+}\left(\lambda z.Max\left(z,\lambda x.\exists e.Theme(x,e)\wedge Travel(e))\right)\right)\right)(y)\right)\\
&=Max\left(x,\lambda y.Max\left(y,\lambda x.\exists e.Theme(x,e)\wedge Travel(e))\right)\right)\\
&=Max\left(x,\lambda y.\exists e.Theme(y,e)\wedge Travel(e))\right)
\end{align*}$$

$$\begin{align*}
Max\left(x,\lambda y.\bigcup\left(\wp^{+}\left(\lambda z.Max\left(z,\bigcup(\wp^{+}(\lambda x.\exists e.Theme(x,e)\wedge Travel(e)))\right)\right)\right)(y)\wedge\exists e.Theme(y,e)\wedge Travel(e)\right)&=Max\left(x,\lambda y.Max\left(y,\bigcup(\wp^{+}(\lambda x.\exists e.Theme(x,e)\wedge Travel(e)))\right)\wedge\exists e.Theme(y,e)\wedge Travel(e)\right)\\
&=Max\big(x,\lambda y.Max\big(y,\lambda x.\exists e.Theme(x,e)\wedge Travel(e)\big)\wedge\exists e.Theme(y,e)\wedge Travel(e)\big)\\
&=Max\big(x,\lambda y.\exists e.Theme(y,e)\wedge Travel(e)\big)
\end{align*}$$

**Result 3** is incoherent:

$$\begin{align*}
Max\left(x,\lambda y.\bigcup\left(\wp^{+}\left(\lambda z.Max\left(z,\bigcup(\wp^{+}(\lambda x.\exists e.Theme(x,e)\wedge Travel(e)))\right)\right)\right)(y)\wedge\exists e.Theme(y,e)\wedge MealService(e)\right)&=Max\left(x,\lambda y.Max\left(y,\bigcup(\wp^{+}(\lambda x.\exists e.Theme(x,e)\wedge Travel(e)))\right)\wedge\exists e.Theme(y,e)\wedge MealService(e)\right)\\
&=Max\left(x,\lambda y.Max\left(y,\bigcup(\wp^{+}(\lambda x.\exists e.Theme(x,e)\wedge MealService(e)))\right)\wedge\exists e.Theme(y,e)\wedge MealService(e)\right)\\
&=Max(x,\emptyset)
\end{align*}$$

Therefore, without $\zeta$, only the counting persons-on-flights reading of (5) is available.
