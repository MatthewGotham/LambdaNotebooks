# Copredication

This lambda notebook contains a demonstration of some ideas found in the following publications:

- Gotham, Matthew (2022). [Property Inheritance, Deferred Reference and Copredication](https://doi.org/10.1093/jos/ffab020). _Journal of Semantics_ 39(1):87–116.
- Gotham, Matthew (2017). [Composing Criteria of Individuation in Copredication](http://academic.oup.com//jos/article/doi/10.1093/jos/ffw008/2555474/Composing-Criteria-of-Individuation-in?guestAccessKey=62fb866e-7bd4-47ae-b6b6-cc1be28a5aeb). _Journal of Semantics_ 34(2):333–371.
- Gotham, Matthew (2014). [Copredication, Quantification and Individuation](https://matthewgotham.github.io/linguistics/MG-thesis.pdf). PhD Dissertation, University College London.

## Basic Examples

```python
%%lamb
||books|| = L x_e: (Books_<e,t>(x), {(x,(L o_(e,e): PHYS_<(e,e),t>(o) & INFO_<(e,e),t>(o)))})
||by_CS_Lewis|| = L x_e: (ByLewis_<e,t>(x), {(x,INFO_<(e,e),t>)})
||on_the_shelf|| = L x_e: (OnShelf_<e,t>(x), {(x,PHYS_<(e,e),t>)})
||MOD|| = L p_<e,(t,{(e,<(e,e),t>)})>: L q_<e,(t,{(e,<(e,e),t>)})>: L x_e: (p(x)[0] & q(x)[0], p(x)[1] | q(x)[1])
||three|| = L p_<e,(t,{(e,<(e,e),t>)})>: L q_<e,(t,{(e,<(e,e),t>)})>: ((Exists x_e: Three_<e,t>(x) & p(x)[0] & q(x)[0] & ~Comp_<(e,<(e,e),t>),t>(x,(L u_(e,e): Exists r_<(e,e),t>: (x,r) << (p(x)[1] | q(x)[1]) & r(u)))), (Set a_(e,<(e,e),t>): p(a[0])[0] & (a << (p(a[0])[1] | q(a[0])[1]))))
```

```python
# Traces and binders
t = [lang.Trace(n) for n in range(5)]
b = [lang.Binder(n) for n in range(5)]
```

```python
counter = 0
def eg(example:str, counter:int) -> None:
    print(f'({counter})\t{example}\n')

def exemplify(example:str, counter:int, compResult) -> None:
    eg(example, counter)
    if len(compResult)<1:
        raise Exception("Composition failed for some reason.")
    print(compResult.source)
    print("")
    if len(compResult)<2:
        display(compResult.content[0].content)
    else:
        for n,cont in enumerate(compResult.content):
            print(f"Result [{n}]:")
            display(cont.content)
```

```python
counter += 1
exemplify("There are three books on the shelf.", counter, ((three * books) * on_the_shelf))
```

We have to do some of the reduction by hand here:

$$
\begin{align*}
\neg Comp(x,&(\lambda u_{(e,e)}.\exists r_{\langle(e,e),t\rangle}.((x,r)\in\{(x,(\lambda o_{(e,e)}.INFO(o)\wedge PHYS(o))),(x,PHYS)\})\wedge r(u))) \\
\Rightarrow\neg Comp(x,&(\lambda u_{(e,e)}.\exists r_{\langle(e,e),t\rangle}.(r\in\{(\lambda o_{(e,e)}.INFO(o)\wedge PHYS(o)),PHYS\})\wedge r(u))) \\
\Rightarrow\neg Comp(x,&(\lambda u_{(e,e)}.(INFO(u)\wedge PHYS(u))\vee PHYS(u))) \\
\Rightarrow\neg Comp(x,&(\lambda u_{(e,e)}.PHYS(u))) \\
\Rightarrow\neg Comp(x,&PHYS)
\end{align*}
$$

```python
counter += 1
exemplify("There are three book by C.S. Lewis.", counter, ((three * books) * by_CS_Lewis))
```

```python
counter += 1
exemplify("There are three books by C.S. Lewis on the shelf.", counter, ((three * ((MOD * by_CS_Lewis) * books)) * on_the_shelf))
```

## Decomposing INFO

```python
%%lamb
||books|| = L x_e: (Books_<e,t>(x), {(x,(L o_(e,e): PHYS_<(e,e),t>(o) & LANG_<(e,e),t>(o)))})
||by_Tolstoy|| = L x_e: (ByTolstoy_<e,t>(x), {(x,PROP_<(e,e),t>)})
||memorized|| = L y_e: L x_e: (Memorize_<(e,e),t>(x,y), {(x,ANI_<(e,e),t>),(y,LANG_<(e,e),t>)})
||Jim|| = Jim_e
```

```python
counter += 1
exemplify("There are three books by Tolstoy.", counter, ((three * books) * by_Tolstoy))
```

```python
counter += 1
exemplify("Jim memorized three books.", counter, ((three * books) * (b[1] * (Jim * (memorized * t[1])))))
```

```python
counter += 1
exemplify("Jim memorized three books by Tolstoy.", counter, ((three * ((MOD * by_Tolstoy) * books)) * (b[1] * (Jim * (memorized * t[1])))))
```
